## **1 Importação de Bibliotecas**

```python
# Importando as bibliotecas necessárias
import pandas as pd

print("Bibliotecas importadas com sucesso!")
```

## **2 Importação da Base de Dados**

```python
# Caminho do arquivo CSV
CAMINHO_ARQUIVO = r"..\data\base_tratada.csv"
print("Caminho do arquivo CSV definido")
```

```python
# Criando o DataFrame Inicial (p/ visulização) a partir do arquivo CSV
df = pd.read_csv(CAMINHO_ARQUIVO, encoding='utf-8')
print("DataFrame Inicial criado com sucesso")

# Visualização do DataFrame Inicial
print("\nExibindo as cinco primeiras linhas do DataFrame Inicial")
df.head()
```

## **3 Preparação da Base de Dados p/ Weka**

```python
# Definição das colunas essenciais de trabalho
sel_atributos = ['Ano', 'Maxima', 'Total', 'NumDiasDeChuva', 'NumDiasSemChuva', 'Estacao', 'Classificacao_Chuva']
print("Atributos selecionados com sucesso")

# Criação do DataFrame Weka
df_weka = df[sel_atributos].copy()
print("\nDataFrame Weka criado com sucesso")
```

```python
# Ajustes de Tipo e Valores
print("Ajustes")
df_weka['Estacao'] = df_weka['Estacao'].replace({'Verão': 'Verao'})
print("   - Valores de 'Estacao' ajustados")

df_weka['Classificacao_Chuva'] = df_weka['Classificacao_Chuva'].replace({
    'Muito Chuvoso': 'MuitoChuvoso',
    'Muito Seco': 'MuitoSeco'
})
print("   - Valores de 'Classificacao_Chuva' ajustados")
```

```python
# Visualização do DataFrame Weka
print("Exibindo as cinco primeiras linhas do DataFrame Weka")
df_weka.head()
```

## **4 Exportação da Base de Dados p/ Weka**

```python
# Definição do caminho e nome do arquivo Weka
CAMINHO_ARQUIVO_WEKA = "../data/base_preparada.arff"
print("Caminho e nome da base Weka definidos com sucesso")
```

```python
# Criação do arquivo ARFF
with open(CAMINHO_ARQUIVO_WEKA, 'w') as f:
    f.write('@RELATION precipitacao_tabatinga\n\n')

    f.write('@ATTRIBUTE Ano NUMERIC\n')
    f.write('@ATTRIBUTE Maxima NUMERIC\n')
    f.write('@ATTRIBUTE Total NUMERIC\n')
    f.write('@ATTRIBUTE NumDiasDeChuva NUMERIC\n')
    f.write('@ATTRIBUTE NumDiasSemChuva NUMERIC\n')
    f.write('@ATTRIBUTE Estacao {Verao,Outono,Inverno,Primavera}\n')
    f.write('@ATTRIBUTE Classificacao_Chuva {MuitoChuvoso,Normal,MuitoSeco}\n')

    f.write('\n@DATA\n')

    for _, row in df_weka.iterrows():
        linha = f"{row['Ano']},{row['Maxima']},{row['Total']},{row['NumDiasDeChuva']},{row['NumDiasSemChuva']},{row['Estacao']},{row['Classificacao_Chuva']}\n"
        f.write(linha)

print(f"Arquivo ARFF gerado com sucesso em: {CAMINHO_ARQUIVO_WEKA}")
```
